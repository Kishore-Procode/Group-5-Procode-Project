# 01. Data Preparation & Inspection - Customer Persona Project

## Project Overview
This notebook covers the **Data Inspection & Cleaning** phase for the Customer Persona Unsupervised Learning project. 
The objective is to inspect raw customer dataset records, validate data quality, verify missing/duplicate values, enforce valid feature ranges, and save the final cleaned dataset for downstream EDA and clustering.

### Key Target Features
- `annual_income_k`: Annual Income of the customer in thousands of dollars ($k).
- `spending_score`: Customer Spending Score assigned by the store based on customer behavior and purchasing patterns (1–100 scale).


## 1. Environment Setup & Data Loading

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os

# Load raw customer data
raw_data_path = "../data/raw/cleaned_customer_data.csv"
print(f"Loading raw dataset from: {raw_data_path}")
df = pd.read_csv(raw_data_path)

# Display first 5 rows
df.head()
```

## 2. Initial Data Inspection

```python
print("Dataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types & Non-Null Counts:")
df.info()
```

```python
print("Descriptive Statistics:")
df.describe()
```

## 3. Data Quality & Validation Checks

```python
# Check missing values
missing_vals = df.isnull().sum()
print("Missing values per column:")
print(missing_vals)

# Check duplicate rows
duplicate_count = df.duplicated().sum()
print(f"\nDuplicate rows found: {duplicate_count}")

# Check value ranges for key features
print("\nMin & Max Values:")
print(f"annual_income_k range: {df['annual_income_k'].min()}k to {df['annual_income_k'].max()}k")
print(f"spending_score range:  {df['spending_score'].min()} to {df['spending_score'].max()}")

# Negative or out-of-bound checks
invalid_income = (df['annual_income_k'] <= 0).sum()
invalid_spending = ((df['spending_score'] < 1) | (df['spending_score'] > 100)).sum()
print(f"Invalid income values (<= 0): {invalid_income}")
print(f"Invalid spending score values (< 1 or > 100): {invalid_spending}")
```

## 4. Summary of Data Cleaning & Export

### Cleaning Findings:
- Total Customer Records: 601
- Missing / Null Values: 0
- Duplicate Records: 0
- Out-of-bounds Values: None (`annual_income_k` within valid positive bounds, `spending_score` within 1-100 range)
- Feature Data Types: Continuous numeric (`float64`)

Since the dataset is clean and structured, it is confirmed ready and exported as the processed dataset `Customer_Data.csv`.


```python
output_dir = "../data/processed"
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, "Customer_Data.csv")

df.to_csv(output_path, index=False)
print(f"Successfully saved clean dataset to: {output_path}")
print("File verification - size:", os.path.getsize(output_path), "bytes")
```
