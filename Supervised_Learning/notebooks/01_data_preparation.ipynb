# 01. Data Preparation & Inspection - Supervised Learning (Loan Approval)

## Project Overview
This notebook handles **Data Inspection & Data Cleaning** for the Loan Approval Prediction project.
The goal is to load raw loan application records, inspect data types and ranges, verify missing or duplicate values, enforce valid business constraints, and export the cleaned dataset for EDA and predictive modeling.

### Features
- `income`: Applicant Annual Income ($).
- `credit_score`: Applicant Credit Score (range 300–850).
- `loan_amount`: Requested Loan Amount ($).
- `employment_years`: Applicant Employment History in years.
- `loan_status`: Binary Target (1 = Approved, 0 = Rejected).


## 1. Environment Setup & Data Loading

```python
import pandas as pd
import numpy as np
import os

raw_data_path = "../data/raw/Loan_Data.csv"
print(f"Loading raw dataset from: {raw_data_path}")
df = pd.read_csv(raw_data_path)
df.head()
```

## 2. Initial Data Inspection

```python
print("Dataset Shape:", df.shape)
print("\nColumns:", df.columns.tolist())
print("\nData Types & Non-Null Counts:")
df.info()
```

```python
print("Descriptive Statistics:")
df.describe()
```

## 3. Data Quality & Constraint Validation

```python
print("Missing values per column:")
print(df.isnull().sum())

print(f"\nDuplicate rows found: {df.duplicated().sum()}")

print("\nTarget Value Counts:")
print(df["loan_status"].value_counts())

# Business constraint checks
print("\nConstraint Checks:")
print("Income <= 0:", (df["income"] <= 0).sum())
print("Credit score outside [300, 850]:", ((df["credit_score"] < 300) | (df["credit_score"] > 850)).sum())
print("Loan amount <= 0:", (df["loan_amount"] <= 0).sum())
print("Employment years < 0:", (df["employment_years"] < 0).sum())
```

## 4. Export Cleaned Dataset

```python
output_dir = "../data/processed"
os.makedirs(output_dir, exist_ok=True)
output_path = os.path.join(output_dir, "cleaned_loan_data.csv")

cleaned_df = df.copy()
cleaned_df.to_csv(output_path, index=False)
print(f"Successfully saved clean dataset to: {output_path}")
print("File size:", os.path.getsize(output_path), "bytes")
```
