```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns


df = pd.read_csv("../data/raw/Loan_Data.csv")
df.head()
```

```python
print("Dataset Shape:", df.shape)
print("\nColumns:")
print(df.columns.tolist())
df.info()
df.describe()
df.isnull().sum()
```

```python
print("Duplicate rows:", df.duplicated().sum())

print("\nUnique values:")
for column in df.columns:
    print(f"{column}: {df[column].unique()}")

print("\nMinimum values:")
print(df.min())

print("\nMaximum values:")
print(df.max())
df.describe()
```

```python
print("Loan status values:")
print(df["loan_status"].value_counts())

print("\nLoan status proportions:")
print(df["loan_status"].value_counts(normalize=True))

print("\nNegative values:")
print((df < 0).sum())

print("Income <= 0:", (df["income"] <= 0).sum())
print("Credit score outside 300-850:", ((df["credit_score"] < 300) | (df["credit_score"] > 850)).sum())
print("Loan amount <= 0:", (df["loan_amount"] <= 0).sum())
print("Employment years < 0:", (df["employment_years"] < 0).sum())
print("Employment years > 50:", (df["employment_years"] > 50).sum())
df.dtypes
```

- Checked for duplicate records.
- Checked for negative values.
- Validated the credit score range.
- Validated loan amount values.
- Validated employment years.
- Checked the target variable (`loan_status`) for valid binary values.
