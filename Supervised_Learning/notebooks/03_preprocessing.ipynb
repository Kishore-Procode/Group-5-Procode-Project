# Loan Approval Prediction — Data Preprocessing

## 1. Overview

This notebook performs the final data preprocessing for the Advanced Loan Approval project.

The cleaned loan dataset generated during the data cleaning stage is used as the input. The purpose of this stage is to validate the cleaned data, separate the input features from the target variable, and prepare the dataset for the machine learning development stage.

The preprocessing work is limited to data preparation. Model-specific feature engineering, model training, model evaluation, and model selection will be performed during the machine learning stage.

---

## 2. Dataset Description

The dataset contains 1,000 loan application records and 5 columns.

The available variables are:

| **Feature**        | **Description**                                        | **Data Type**    | **Role**        |
| ------------------ | ------------------------------------------------------ | ---------------- | --------------- |
| `income`           | Annual income of the loan applicant                    | Integer          | Input Feature   |
| `credit_score`     | Credit score of the loan applicant                     | Integer          | Input Feature   |
| `loan_amount`      | Amount of loan requested by the applicant              | Integer          | Input Feature   |
| `employment_years` | Number of years the applicant has been employed        | Integer          | Input Feature   |
| `loan_status`      | Loan approval outcome (`0` = Rejected, `1` = Approved) | Integer / Binary | Target Variable |


The target variable is `loan_status`.

The target values are:

- `0` — Rejected
- `1` — Approved
