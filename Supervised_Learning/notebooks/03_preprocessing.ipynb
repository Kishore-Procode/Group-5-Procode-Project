# 03. Preprocessing & Feature Engineering

## Overview
This notebook handles **Data Preprocessing & Scaling** before feeding features into supervised classification algorithms.

### Objectives:
1. Feature/Target separation.
2. Train/Test Data Split (80/20 stratified split).
3. Feature Scaling using `StandardScaler`.
4. Summary documentation for ML pipeline readiness.


## 1. Environment Setup & Data Loading

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("../data/processed/cleaned_loan_data.csv")
print("Loaded dataset shape:", df.shape)
df.head()
```

## 2. Feature & Target Matrix Separation

```python
X = df.drop(columns=["loan_status"])
y = df["loan_status"]

print("Feature Columns:", X.columns.tolist())
print("Target Column: loan_status")
print("X shape:", X.shape, "| y shape:", y.shape)
```

## 3. Stratified Train-Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

print(f"Train set: {X_train.shape[0]} samples")
print(f"Test set : {X_test.shape[0]} samples")
print("\nTarget distribution in Train set:")
print(y_train.value_counts(normalize=True))
```

## 4. Feature Scaling (`StandardScaler`)

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

df_train_scaled = pd.DataFrame(X_train_scaled, columns=X.columns)
print("Scaled Train Features Summary:")
df_train_scaled.describe().round(3)
```
